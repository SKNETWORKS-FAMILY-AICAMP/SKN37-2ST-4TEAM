```python
import joblib
import pickle

# 1번부터 5번 파일 경로 목록
input_files = ["./model/machine_learning/decision_tree_v1.pkl", 
               "./model/machine_learning/knn_v1.pkl", 
               "./model/machine_learning/naive_bayes_v1.pkl", 
               "./model/machine_learning/random_forest_v1.pkl", 
               "./model/machine_learning/xgboost_v1.pkl"]

# 6번 파일 형태로 저장할 출력 파일 경로 목록
output_files = ["./model/machine_learning/decision_tree_v1.pkl", 
               "./model/machine_learning/knn_v1.pkl", 
               "./model/machine_learning/naive_bayes_v1.pkl", 
               "./model/machine_learning/random_forest_v1.pkl", 
               "./model/machine_learning/xgboost_v1.pkl"]

for in_path, out_path in zip(input_files, output_files):
    try:
        # 1. 기존 pickle/joblib 모델 파일 로드
        model = joblib.load(in_path)
        
        # 2. 6번 파일과 같은 joblib 직렬화 형태(바이너리/텍스트)로 저장
        joblib.dump(model, out_path)
        print(f"변환 완료: {in_path} -> {out_path}")
        
    except Exception as e:
        # pickle 라이브러리로 대체 시도
        try:
            with open(in_path, "rb") as f:
                model = pickle.load(f)
            joblib.dump(model, out_path)
            print(f"pickle 대체 변환 완료: {in_path} -> {out_path}")
        except Exception as err:
            print(f"파일 {in_path} 변환 실패: {err}")
```
